## Issue 1 — What does a downloaded financial-data row mean, and which request settings determine it?

A date and number do not yet identify a financial observation. Interpreting a returned row requires the access library, data provider, security, venue, field, unit, observation frequency, and meaning of its date or timestamp. The request must also record the ticker, inclusive start boundary, exclusive end boundary, interval, and explicitly selected adjustment options. Query settings show what was requested; returned dates, fields, and values show what the provider response contained.

For an inclusive start date $s$, exclusive end date $e$, and set of returned dates $D_{\mathrm{returned}}$, a correctly bounded response satisfies

$$
D_{\mathrm{returned}}\subseteq[s,e)=\{d:s\leq d<e\}.
$$

This subset relation does not require every calendar date inside the interval to appear. Weekends, exchange holidays, unavailable observations, and failed requests can reduce the returned set. Even when every returned date is inside the interval, the check does not prove that the provider's values or field construction are correct.

## Issue 2 — Why is a ticker not enough to identify an asset?


A ticker is a trading label, not a permanent asset identifier. Using ticker as the only database key creates two opposite errors:

| Case | What the records show | Error caused by using ticker alone |
|---|---|---|
| Same ticker, different assets | `XYZ` matches P003 and P004 | A price may be assigned to the wrong asset |
| Same asset, different tickers | Meta Class A common stock changes from `FB` to `META` | One price history may be split into two unrelated assets |

Use the provider asset identifier to identify the asset within that data source. Store the ticker together with the venue and the dates during which the ticker applies. If these records do not identify one asset, do not assign the price or calculate that asset's return. This issue needs record matching rather than a financial equation.

## Issue 3 — What exactly do Open, High, Low, Close, and Volume summarize?


OHLCV compresses many observations into five values for one interval. In the basic construction used in Code 1, Open is the first included price, High is the largest, Low is the smallest, Close is the last, and Volume is the sum of the included quantities. A real provider may apply exchange rules, auction rules, session boundaries, or transaction filters when constructing these fields. Therefore, the five column names do not by themselves define which observations are included.

### Symbols and units

Let $p_{t,j}$ be the $j$th included price in time order within interval $t$, $q_{t,j}$ its included number of shares, and $n_t$ the number of included artificial trades. The prices use artificial currency units per share, and the quantities use shares. All rows must describe the same artificial asset and interval.

### Reasoning and formulation

$$
\begin{aligned}
O_t &= p_{t,1},\\
H_t &= \max_{1\leq j\leq n_t} p_{t,j},\\
L_t &= \min_{1\leq j\leq n_t} p_{t,j},\\
C_t &= p_{t,n_t},\\
V_t &= \sum_{j=1}^{n_t} q_{t,j}.
\end{aligned}
$$

The ordering determines Open and Close; the price values determine High and Low; the quantities determine Volume. These definitions imply the following inequality:

$$
L_t
\leq
\min(O_t,C_t)
\leq
\max(O_t,C_t)
\leq
H_t.
$$

A row that fails this inequality cannot be produced by the stated construction. A row that satisfies it is not automatically correct: the inequality cannot determine the asset, trading hours, eligible transaction categories, or Volume unit. If different assets or intervals are mixed, the equations still produce numbers but those numbers do not summarize one valid interval.

## Issue 4 — Which field matches the return being measured?


A return measures the change in one defined series. A closing price, adjusted close, and total-return index can represent different financial changes even when each uses the same return formula. Select the series that answers the financial question and retain its documented definition. A series is not preferable merely because it produces the largest historical return.

### Symbols and units

Let $F_{t-1}$ and $F_t$ be values from the same series definition, asset, currency, unit, share basis when applicable, and stated interval. Let $r_t^{(F)}$ be the decimal return calculated from that series. A return is dimensionless: 0.01 means 1%. The beginning value must be nonzero; a price input used here must also be positive.

### Reasoning and derivation

First define the change in the selected series:

$$
\Delta F_t = F_t-F_{t-1}.
$$

Divide that change by the beginning field value and substitute the definition of $\Delta F_t$:

$$
\begin{aligned}
r_t^{(F)}
&=\frac{\Delta F_t}{F_{t-1}}\\
&=\frac{F_t-F_{t-1}}{F_{t-1}}\\
&=\frac{F_t}{F_{t-1}}-1,
\qquad F_{t-1}\neq 0.
\end{aligned}
$$

The algebra is the same for each series, but the financial meaning comes from its definition. Applying the calculation to `Close` produces a closing-price return. Applying it to `Adj Close` or a total-return index produces the return calculated from that documented series. The formula alone cannot reveal which corporate actions or distributions were incorporated.

## Issue 5 — How can a stock split change price and share count without changing position value?


A stock split changes how one ownership position is divided into shares.

### Symbols and units

Let $s>0$ be the number of new shares received for each old share, $N_{t-1}$ the pre-split quantity in shares, and $P_{t-1}$ the pre-split price in currency units per share. Position value $V_{t-1}=N_{t-1}P_{t-1}$ is measured in currency units.

### Reasoning and derivation

If the split is the only change, the new share quantity and the price per share implied by the ratio are

$$
\begin{aligned}
N_t &= sN_{t-1},\\
P_t &= \frac{P_{t-1}}{s}.
\end{aligned}
$$

Multiplying price by quantity shows why the position value is unchanged under these restricted conditions:

$$
\begin{aligned}
V_t
&=N_tP_t\\
&=(sN_{t-1})\left(\frac{P_{t-1}}{s}\right)\\
&=N_{t-1}P_{t-1}\\
&=V_{t-1}.
\end{aligned}
$$

The price-per-share change caused by the ratio alone is $1/s-1$, while the position-value return caused by the ratio alone is zero. These equations compare the old share quantity and price with the new share quantity and price. They are not instructions to divide a provider's already adjusted price a second time.

Actual market prices can also move across the event. Therefore, the ratio calculation does not replace an observed-return calculation, and an observed price decline does not by itself prove that a split occurred.


### Before Code 1


For an artificial two-for-one split, begin with one share priced at 100 artificial currency units per share. Assume that the split is the only change. Predict the after-split price per share, share quantity, and position value. Then predict the percentage change in each quantity.

````python
split_ratio = 2.0
before = pd.Series(
    {"Price per share": 100.0, "Share quantity": 1.0, "Position value": 100.0}
)
after = pd.Series(
    {"Price per share": before["Price per share"] / split_ratio,
     "Share quantity": before["Share quantity"] * split_ratio}
)
after["Position value"] = after["Price per share"] * after["Share quantity"]
split_summary = pd.DataFrame({"Before": before, "After": after})
split_summary["Percent change"] = (split_summary["After"] / split_summary["Before"] - 1) * 100
split_summary.insert(
    0, "Unit", ["artificial currency units per share", "shares", "artificial currency units"]
)
display(split_summary.style.format(
    {"Before": "{:.2f}", "After": "{:.2f}", "Percent change": "{:.2f}"}
).set_caption(
    "Artificial two-for-one split with no other price movement, fees, or taxes"
))
assert np.allclose(split_summary["Percent change"], [-50.0, 100.0, 0.0])
````

## Issue 6 — How does a cash dividend affect holding-period return?


A **price return** measures the price change. A **holding-period return** measures the change in the value of the investment, including income earned while holding it. First establish whether the shares are eligible for a dividend. Then distinguish the ex-dividend date, when the right to this regular dividend belongs to the eligible holder, from the payable date, when the announced payment is due.

An eligible investor can reach the end of a measurement period before payment arrives. The dividend is then a **dividend receivable**: an amount owed to the investor but not yet received as cash. The [GIPS Standards Handbook for Firms, Provision 2.B.3](https://www.gipsstandards.org/standards/gips-standards-for-firms/gips-standards-handbook-for-firms/) recommends recognizing dividends on the ex-dividend date rather than waiting for payment. This lesson uses that accrual treatment for its worked examples; it does not claim compliance with the full standards.

### From investment value to return

Consider a fixed quantity $N>0$ of shares held throughout an interval that begins before the ex-dividend date. The worked example uses $N=1$ share. The beginning price is $P_0>0$ currency units per share. There is no beginning cash balance or dividend receivable, so beginning wealth is $W_0=NP_0$ currency units.

At the ending date, $P_1$ is the stock price per share. Let $C_1$ be dividend cash already received and retained, and $A_1$ the value of the unpaid dividend receivable, both in currency units for the whole position. The example assumes no split, reinvestment, external deposit or withdrawal, interest, tax, fee, or currency conversion. It values the receivable at face value, ignoring discounting and payment risk. Therefore,

$$
W_1=NP_1+C_1+A_1.
$$

Let $D_1$ be the dividend earned per eligible share in currency units per share, so $ND_1=C_1+A_1$. Each dividend appears in cash or in the receivable, never both. The price return is

$$
r_{\mathrm{price}} = \frac{P_1-P_0}{P_0}=\frac{P_1}{P_0}-1.
$$

Substitute ending wealth into the holding-period return:

$$
\begin{aligned}
r_H
&=\frac{W_1-W_0}{W_0}\\
&=\frac{NP_1+C_1+A_1-NP_0}{NP_0}\\
&=\frac{P_1+D_1}{P_0}-1.
\end{aligned}
$$

Subtracting the first equation from the second shows exactly how the included cash dividend changes the result:

$$
r_H-r_{\mathrm{price}}=\frac{D_1}{P_0}.
$$

If a later measurement interval begins with a receivable already present, include it in beginning wealth too. Receiving that cash changes the composition of wealth; it is not a second dividend gain. The general calculation remains $(W_1-W_0)/W_0$ under the no-external-flow assumption.


### Controlled example

Code 1 uses artificial values: a beginning price of 100 currency units per share, an ending price of 98 currency units per share, and an earned dividend of 3 currency units for one eligible share.
````python
beginning_price = 100.0  # Currency units per share.
ending_price = 98.0
share_quantity = 1.0
beginning_wealth = share_quantity * beginning_price
earned_dividend = share_quantity * 3.0  # 3 currency units per eligible share.
price_return = ending_price / beginning_price - 1

controlled_dividend_returns = pd.DataFrame({
    "Endpoint under the stated assumptions": ["Before payment", "After payment"],
    "Stock value (currency units)": [share_quantity * ending_price] * 2,
    "Dividend cash retained (currency units)": [0.0, earned_dividend],
    "Dividend receivable (currency units)": [earned_dividend, 0.0],
})
controlled_dividend_returns["Total investment value (currency units)"] = (
    controlled_dividend_returns["Stock value (currency units)"]
    + controlled_dividend_returns["Dividend cash retained (currency units)"]
    + controlled_dividend_returns["Dividend receivable (currency units)"]
)
controlled_dividend_returns["Holding-period return (%)"] = 100 * (
    controlled_dividend_returns["Total investment value (currency units)"]
    / beginning_wealth - 1
)
display(controlled_dividend_returns.style.format({
    "Stock value (currency units)": "{:.2f}",
    "Dividend cash retained (currency units)": "{:.2f}",
    "Dividend receivable (currency units)": "{:.2f}",
    "Total investment value (currency units)": "{:.2f}",
    "Holding-period return (%)": "{:.2f}",
}).hide(axis="index").set_caption(
    "Artificial eligible one-share holding: beginning wealth 100 currency units; "
    "stock price return -2%; no costs, taxes, reinvestment, or price change at payment"
))

assert np.isclose(price_return, -0.02)
assert np.allclose(
    controlled_dividend_returns["Total investment value (currency units)"], 101.0
)
assert np.allclose(controlled_dividend_returns["Holding-period return (%)"], 1.0)
assert np.isclose(
    controlled_dividend_returns["Total investment value (currency units)"].iloc[1]
    / controlled_dividend_returns["Total investment value (currency units)"].iloc[0] - 1,
    0.0,
)
````

### Why the adjusted-price ratio can differ from stock plus dividend value

Issue 4 showed different returns from `Close` and `Adj Close`. Now compare the adjusted-price ratio with the holding-period formula just derived, keeping the same asset, two dates, prices, and dividend.

[Yahoo Help](https://help.yahoo.com/kb/SLN28256.html) describes a dividend multiplier based on the dividend divided by the preceding closing price. Consider one dividend $D$ with $0\leq D<P_0$, no split between the endpoints, and compatible per-share units. The multiplier for earlier prices is

$$
m=1-\frac{D}{P_0}.
$$

Ignoring later adjustment factors that multiply both endpoints equally and cancel in their ratio, the adjusted beginning price is $mP_0=P_0-D$. The adjusted-price return implied by this method is

$$
r_{\mathrm{adj}}
=\frac{P_1}{mP_0}-1
=\frac{P_1}{P_0-D}-1.
$$

The eligible investor's stock-plus-dividend value return instead uses the original beginning investment:

$$
r_H=\frac{P_1+D}{P_0}-1.
$$

These are not algebraically identical:

$$
\begin{aligned}
r_{\mathrm{adj}}-r_H
&=\frac{P_1}{P_0-D}-\frac{P_1+D}{P_0}\\
&=\frac{D(P_1+D-P_0)}{P_0(P_0-D)}\\
&=\frac{D}{P_0-D}\,r_H.
\end{aligned}
$$

The difference depends on the dividend and the holding-period return; it is not always positive. In the artificial 100-to-98 example with a dividend of 3 currency units per share, stock plus dividend gives 1.0000%, while the multiplier method gives $98/97-1$, approximately 1.0309%. The difference follows from the two constructions, not from an extra dividend earned by the investor.

## Issue 7 — Which unit and currency does each rate or return use?


A correct number can communicate the wrong financial result when its unit is missing or incorrect. A decimal value of 0.01, a return of 1%, and 100 basis points represent the same proportional amount. One basis point is 0.0001 in decimal form, or 0.01 percentage point.

Every financial number must identify its unit. In a sentence, place the ISO currency code before a monetary amount, as in `USD 1.20`. In a table or figure, put the unit in the column header, axis label, or caption so that every displayed value can be interpreted without repeating the unit in every cell.

The following expressions are illustrative notation, not market observations:

| Financial quantity | Complete expression | Meaning |
|---|---:|---|
| Cash amount | USD 1.20 | 1.20 U.S. dollars |
| Share price | USD 125.40 per share | 125.40 U.S. dollars for one share |
| Return | 1.20% | A proportional change of 0.0120 |
| Interest-rate change | 25 basis points | A change of 0.25 percentage point |
| Trading volume | 1,200 shares | The number of shares represented by the value |
| Exchange rate | TWD 32.00 per USD | 32.00 New Taiwan dollars for one U.S. dollar |

Percentage points measure the arithmetic difference between two rates. A relative percentage change divides that difference by the starting rate. Let $i_0$ and $i_1$ be beginning and ending rates written as decimals. The arithmetic change is

$$
\Delta i=i_1-i_0.
$$

The same arithmetic change can be reported in percentage points or basis points:

$$
\text{percentage-point change}=100\Delta i,
$$

$$
\text{basis-point change}=10{,}000\Delta i.
$$

When $i_0>0$, the relative change is

$$
\text{relative change}=\frac{i_1-i_0}{i_0}.
$$

These quantities answer different questions. If a rate rises from 4% to 5%, the arithmetic increase is 1 percentage point, or 100 basis points, while the relative increase is 25%.

### How much is the same stock worth in New Taiwan dollars?

A stock can rise in U.S. dollars while its value in New Taiwan dollars falls. To measure the return for an investor who reports wealth in TWD, first convert the beginning and ending stock values into TWD. A return is dimensionless, but its numerical value depends on the currency used to value the investment.

Consider one artificial USD-denominated stock. Its price rises from USD 100.00 per share to USD 102.00 per share. At the beginning, one U.S. dollar exchanges for TWD 32.00; at the end, it exchanges for TWD 28.80. These prices and exchange rates are teaching assumptions, not observations of AAPL, MSFT or an actual currency market.

Use two matched valuation instants, labeled beginning and end. At each instant, the stock price and exchange rate refer to that same instant; matching calendar dates alone would not establish this for real observations. Hold one share throughout, with no dividends, splits, trades, deposits or withdrawals. Ignore conversion fees, bid–ask spreads, taxes and hedging. The calculation measures the stock's value in TWD, not a broker's realized conversion proceeds.

Let $P_t$ be the USD price of one share and $X_t$ the TWD value of one USD. Let $N$ be the fixed number of shares and $W_t$ the holding's value in TWD. Here $t=0$ means beginning and $t=1$ means end. Both prices and exchange rates are positive, and $N>0$.

| Symbol | Quantity | Unit |
|---|---|---|
| $P_t$ | Stock price at instant $t$ | USD per share |
| $X_t$ | TWD received per USD at the assumed valuation rate | TWD per USD |
| $N$ | Shares held throughout the interval | shares |
| $W_t$ | Stock holding's value at instant $t$ | TWD |

Multiplication converts the holding into TWD because the units cancel:

$$
W_t=NP_tX_t,
\qquad
\text{shares}\times\frac{\text{USD}}{\text{share}}
\times\frac{\text{TWD}}{\text{USD}}=\text{TWD}.
$$

The stock's USD return is $r_L=P_1/P_0-1$. The USD currency return measured in TWD is $r_{FX}=X_1/X_0-1$. Both are decimal returns over the same interval. A negative $r_{FX}$ means that one USD buys fewer TWD at the end. The subscript $L$ refers to the stock's price currency, USD, not the location of the company's business activities.

The TWD holding return $r_R=W_1/W_0-1$ follows from those two valuations:

$$
\begin{aligned}
1+r_R
&=\frac{W_1}{W_0}
=\frac{NP_1X_1}{NP_0X_0}\\
&=\frac{P_1}{P_0}\frac{X_1}{X_0}
=(1+r_L)(1+r_{FX}),\\
r_R&=r_L+r_{FX}+r_Lr_{FX}.
\end{aligned}
$$

The product arises because both the stock price and the TWD value of each dollar change over the **same** holding interval. It is not a calculation of returns in two consecutive periods. Merely adding the two returns omits $r_Lr_{FX}$.

The [CFA Institute's public currency-management summary](https://www.cfainstitute.org/insights/professional-learning/refresher-readings/2026/currency-management-introduction) describes this multiplicative relation between foreign-asset and currency returns. The numerical examples here are our own teaching calculations. Real exchange-rate data also require a stated quotation direction and observation time. For example, the [ECB's reference-rate page](https://www.ecb.europa.eu/stats/policy_and_exchange_rates/euro_reference_exchange_rates/html/index.en.html) states its base currency and warns that its reference rates are for information, not transaction use; it is not the source of these TWD/USD assumptions.

### What changes if the exchange rate is quoted the other way around?

Reversing the quote changes the number and its unit, not the underlying holding. Let $Y_t=1/X_t$ be USD per TWD, using the same ideal valuation rate with no bid–ask spread. Since the stock value is in USD, divide it by $Y_t$ to express it in TWD:

$$
W_t=\frac{NP_t}{Y_t},
\qquad
\frac{\text{USD}}{\text{USD}/\text{TWD}}=\text{TWD}.
$$

Let $r_Y=Y_1/Y_0-1$ be the decimal return calculated from this inverse quote. Its relation to $r_{FX}$ is exact:

$$
\begin{aligned}
1+r_Y&=\frac{1/X_1}{1/X_0}
=\frac{1}{1+r_{FX}},\\
r_Y&=\frac{-r_{FX}}{1+r_{FX}},\\
1+r_R&=\frac{P_1/P_0}{Y_1/Y_0}
=\frac{1+r_L}{1+r_Y}.
\end{aligned}
$$

The percentage change of an inverse quote is **not** generally the negative of the original percentage change. Its denominator also changes. For real dealer quotes, bid and ask prices must be distinguished; reciprocating one quoted price does not identify the price at which a customer can trade in the opposite direction.

## Issue 8 — How do simple and log returns describe the same holding interval?

A **simple return** reports the proportional price change from beginning price $P_s$ to ending price $P_e$. Here $s<e$ are the same two valuation endpoints for both calculations. Use finite positive prices for the same asset, field, currency and per-share basis, with no distribution added. Prices have currency-per-share units; both returns below are dimensionless decimals.

Divide the price gain by the beginning price:

$$
r_{s,e}=\frac{P_e-P_s}{P_s}=\frac{P_e}{P_s}-1,
\qquad P_s>0.
$$

A **log return** takes the natural logarithm of the same ending-to-beginning price ratio. Its symbol is $g_{s,e}$. Since $1+r_{s,e}=P_e/P_s$, substitution gives

$$
g_{s,e}=\ln\!\left(\frac{P_e}{P_s}\right)=\ln(1+r_{s,e}),
\qquad P_s>0,\;P_e>0.
$$

The transformations are exact:

$$
g_{s,e}=\ln(1+r_{s,e}),
\qquad
r_{s,e}=e^{g_{s,e}}-1.
$$

The logarithm also gives $g_{s,e}=\ln P_e-\ln P_s$ when both prices use the same units. The next issue shows how these differences cancel at shared endpoints. A zero ending value corresponds to a -100% simple return, but not a finite log return; the positive-price examples here stop before attempting that logarithm.

A simple return of 0.01 means a 1% price change. Multiplying a log return by 100 changes only its display scale; it does not turn the value into a simple percentage return. For a small return the two numbers can be close, but they are not interchangeable. Campbell, Lo, and MacKinlay's chapter on [prices, returns, and compounding](https://doi.org/10.1515/9781400830213-005) provides the standard return definitions used here.

## Issue 9 — How do adjacent simple returns produce a full-interval return and wealth path?


Issue 8 defined a simple return for one pair of price endpoints. This issue asks when several such returns can be combined into one complete holding interval. Let $P_0,P_1,\ldots,P_T$ be ordered, compatible prices for one asset, and let

$$
r_t=\frac{P_t}{P_{t-1}}-1.
$$

The gross return for subperiod $t$ is $1+r_t=P_t/P_{t-1}$. When every ending endpoint is also the beginning endpoint of the next subperiod, the intermediate prices cancel:

$$
\begin{aligned}
1+R_{0,T}
&=\prod_{t=1}^{T}(1+r_t)\\
&=\frac{P_1}{P_0}\frac{P_2}{P_1}\cdots\frac{P_T}{P_{T-1}}\\
&=\frac{P_T}{P_0}.
\end{aligned}
$$

Therefore,

$$
R_{0,T}=\prod_{t=1}^{T}(1+r_t)-1=\frac{P_T}{P_0}-1.
$$

The corresponding log return for subperiod $t$ is $g_t=\ln(1+r_t)$. For the same ordered chain of positive price endpoints,

$$
\sum_{t=1}^{T}g_t
=\ln\!\left(\frac{P_T}{P_0}\right),
\qquad
R_{0,T}=\exp\!\left(\sum_{t=1}^{T}g_t\right)-1.
$$

This exact addition is one reason log returns are often convenient in research that decomposes one asset's return over adjacent intervals. The result still requires compatible endpoints, and the summed log return must be transformed back before it is reported as a simple full-interval return.

The same product generates a wealth path from one initial unit:

$$
W_0=1,
\qquad
W_t=\prod_{j=1}^{t}(1+r_j).
$$

The terminal return $W_T-1$ does not contain the intermediate values $W_1,\ldots,W_{T-1}$. Reordering the same subperiod returns leaves their final product unchanged, but it can change the wealth observed after the earlier subperiods. This is a rearrangement of the same numerical inputs, not permission to reorder a real price history.

Adding simple returns generally gives a different result. For two subperiods,

$$
(1+r_1)(1+r_2)-1=r_1+r_2+r_1r_2.
$$

The product term $r_1r_2$ is omitted when the two returns are simply added. Compounding can support a full-interval claim only when the returns use the same asset, field, currency, unit, adjustment treatment, and one ordered chain of endpoints with no missing link or overlap. The algebraic identity remains true for that compatible price series. Interpreting the compounded price return as an investor's change in wealth additionally requires the stated treatment of distributions, external contributions or withdrawals, transaction costs, fees, and taxes. A closing-price path does not include a cash distribution unless the selected return measure adds or adjusts for it.

## Issue 10 — Which average describes one-period returns and which reproduces compounded wealth?


Issue 9 combined a realized sequence of adjacent returns. This issue asks a different question: if that sequence must be summarized by one return per period, which average matches the intended statement?

Let $r_t$ be the observed simple return for period $t$, expressed as a decimal, and let $T$ be the number of periods. Use one declared period convention throughout, such as consecutive daily closes or consecutive month-end closes. Daily close-to-close intervals can include a weekend; they do not all contain the same number of calendar hours. Do not pool daily and monthly observations into one equally weighted average.

The arithmetic mean return adds the $T$ observed returns and divides by their count:

$$
\bar r_{\mathrm{A}}=\frac{1}{T}\sum_{t=1}^{T}r_t.
$$

It describes the average of the observed one-period returns. It does not generally reproduce the wealth produced by investing through all $T$ periods.

For the same observed sequence, start with one unit of wealth and compound the returns without external deposits or withdrawals. The geometric mean return is the constant per-period return that reproduces the same terminal wealth. Equating that constant-return path to the observed product and taking the $T$th root gives

$$
\begin{aligned}
(1+\bar r_{\mathrm{G}})^T
&=\prod_{t=1}^{T}(1+r_t),\\
\bar r_{\mathrm{G}}
&=\left[\prod_{t=1}^{T}(1+r_t)\right]^{1/T}-1.
\end{aligned}
$$

When every $r_t>-1$, the same calculation can be written as

$$
\bar r_{\mathrm{G}}
=\exp\left(\frac{1}{T}\sum_{t=1}^{T}\ln(1+r_t)\right)-1.
$$

The arithmetic–geometric mean inequality applied to the positive gross returns gives

$$
\bar r_{\mathrm{G}}\leq\bar r_{\mathrm{A}},
$$

with equality only when all observed returns are equal. A return of exactly -100% produces zero terminal wealth and an equivalent constant return of -100%; the code below uses the logarithmic expression and therefore requires every return to be greater than -100%.

Neither result below is annualized. The [average-return section in Eric Zivot's finance text](https://bookdown.org/compfinezbook/introFinRbook/Return-Calculations.html#average-returns) supports the distinction between the arithmetic average and the constant rate that reproduces compounded wealth.

### When can a historical average estimate expected return?

An expected return averages possible outcomes using their probabilities. It is not a promise about the next outcome. Capital $R_t$ denotes an uncertain simple return; lowercase $r_t$ denotes its observed value. Let $\mu$ be the same finite expected return in every sampled period. All returns use the same asset, currency, period convention and dividend treatment. Use fixed observation periods, not dates selected because their returns were high.

The sample arithmetic mean can estimate this $\mu$. Before observing the sample, its expectation is

$$
\begin{aligned}
\widehat{\mu}&=\frac{1}{T}\sum_{t=1}^{T}R_t,\\
\mathbb{E}[\widehat{\mu}]
&=\frac{1}{T}\sum_{t=1}^{T}\mathbb{E}[R_t]
=\frac{1}{T}\sum_{t=1}^{T}\mu
=\mu.
\end{aligned}
$$

After observing the returns, the numerical estimate is $\widehat{\mu}=\bar r_{\mathrm{A}}$. This is an unbiased estimator under the stated common-mean assumption: its average over possible samples equals $\mu$, not necessarily its value in any one sample. Independence and a normal distribution are not needed for this expectation identity. Dependence still matters when measuring estimation uncertainty.

Using this estimate for a future period requires an additional assumption that the future mean remains the same. It does not establish the expected return given today's particular news or market conditions. The geometric mean reproduces observed terminal wealth, not every intermediate value; it is not generally an estimate of the mean simple return $\mu$.

For the estimator/estimate distinction and the linearity-of-expectation derivation, see Zivot's [Section 7.1 and Section 7.5.1](https://bookdown.org/compfinezbook/introFinRbook/Estimation-of-The-GWN-Model.html). That chapter works with a Gaussian model; only the common-mean identity is used here, not its full distributional assumptions or inference procedures.

## Issue 11 — Do the date labels produce one correctly ordered row per intended observation?


A return calculation needs dates that are parsed with the source's stated format, placed in chronological order, and checked for repeated observations. Sorting unparsed text can give the wrong order. $$

## Issue 12 — Which prices or returns require further review?


A data check should identify the condition it found without deciding that the observation is wrong. This issue keeps five conditions separate:

| Condition | Concrete example | Calculation or observation | What the condition establishes |
|---|---:|---:|---|
| Missing price | `100, missing, 102` currency units per share | Neither adjacent return can be calculated with `fill_method=None` | A required numeric endpoint is absent |
| Unchanged price | 100 to 100 currency units per share | Simple return = 0% | Two reported prices are equal; the reason is still unknown |
| Zero ending price | 100 to 0 currency units per share | Simple return = -100%; log return is not finite | The position loses all price value under these endpoints |
| Zero beginning price | 0 to 150 currency units per share | The software displays `inf`; the financial return is undefined | The simple-return denominator is invalid |
| Large finite return | 149 to 210 currency units per share | Simple return = 40.94% | The finite return exceeds the declared review threshold |

A negative common-stock price, a zero used as a placeholder, a stale repeated value, a corporate action, and a genuine large market move require different responses. The code therefore never replaces, deletes, forward-fills, winsorizes, or clips a value automatically. $$
|r_t|>\tau
\quad\text{and}\quad
r_t\text{ is finite}.
$$

The strict greater-than rule matters: a return exactly equal to $\tau$ is not marked. The threshold $\tau=0.25$, or 25%, is a classroom review setting, not a financial definition of an error.

## Issue 14 — Which instant and data interval does a timestamp represent?


A timestamp such as `2026-01-05 16:00` gives a clock reading but does not identify one instant until its time zone is known. In pandas, `tz_localize` assigns a time zone to a time-zone-naive clock reading. After an instant has been established, `tz_convert` changes its displayed time zone while preserving the instant. Treating a UTC clock reading as New York time changes the instant; conversion cannot repair that earlier assumption.

Knowing the named zone is necessary but may not be sufficient near a daylight-saving change. When clocks move back, a local clock reading can occur twice; when clocks move forward, some readings never occur. Do not choose an offset by guessing. Use the source's UTC timestamp or explicit offset to resolve the instant. The examples below use dates and clock readings that do not have those problems; [pandas documents these localization cases](https://pandas.pydata.org/docs/user_guide/timeseries.html#time-zone-handling).

UTC is the reference time zone with offset `+00:00`; it does not move its clocks for daylight saving time. A named time zone such as `America/New_York` contains rules that determine whether its offset is -5 or -4 hours on a particular date. The offset describes one date and instant; it is not a complete replacement for the named time zone.

Let $c_{\mathrm{local}}$ be the local clock time and $u(t)$ the UTC offset in hours on date $t$. The UTC clock time satisfies

$$
c_{\mathrm{UTC}}=c_{\mathrm{local}}-u(t).
$$

The sign describes the local clock relative to UTC. `UTC+08:00` means the local clock is eight hours ahead of UTC, so 5:00 a.m. in Taipei converts to 9:00 p.m. UTC on the preceding date. `UTC-05:00` means the local clock is five hours behind UTC, so 4:00 p.m. in New York converts to 9:00 p.m. UTC on the same date.

New York's offset is not constant throughout the year. A 4:00 p.m. New York timestamp is 9:00 p.m. UTC when the offset is -5 hours and 8:00 p.m. UTC when the offset is -4 hours.

## Issue 15 — How are asset returns aligned and combined into a portfolio return?

A portfolio return must reconcile with the change in the total value of its holdings. Before adding asset gains, make sure every component describes the same holding interval. For asset $j$, let $P_{j,s}$ and $P_{j,e}$ be finite, positive prices at beginning endpoint $s$ and ending endpoint $e$. With a consistent per-share basis and no distributions, its decimal simple price return is

$$
r_{j;s,e}=\frac{P_{j,e}}{P_{j,s}}-1.
$$

Matching only an ending-date label is insufficient. Every asset must use the same $s$ and $e$, compatible price fields and corporate-action conventions, and nonmissing endpoint prices. Convert holding values into one reporting currency before adding them, as in Issue 7. The examples below already use one currency and matching valuation conventions; the artificial example uses calendar-date labels for its stated closing observations.

### Why do beginning weights determine the portfolio return?

The portfolio's gain is the sum of the gains on the assets actually held. Each asset's return applies to the money invested in that asset at the **beginning** of the interval. A 70% weight means 70% of beginning portfolio value, not 70% of the total number of shares.

For this derivation, hold fixed share quantities from $s$ to $e$, with no trades, distributions, splits, deposits or withdrawals inside the interval. Prices are finite, positive and in one reporting currency per share. The portfolio is fully invested in these assets, with nonnegative holdings and no borrowing, residual cash, fees or taxes. Fractional shares are allowed as a teaching assumption. These conditions describe the calculation, not an available brokerage service.

| Symbol | Meaning | Unit |
|---|---|---|
| $J$ and $j$ | Number of assets and asset index | count and index |
| $N_j$ | Shares of asset $j$ held throughout the interval | shares |
| $V_{j,s}$, $V_{j,e}$ | Beginning and ending value of that holding | reporting-currency units |
| $W_s$, $W_e$ | Beginning and ending total portfolio value | reporting-currency units |
| $w_{j,s}$ | Fraction of beginning portfolio value invested in asset $j$ | decimal weight |
| $r_{j;s,e}$, $r_{p;s,e}$ | Asset and portfolio simple return over $s$ to $e$ | decimal return |
| $c_{j;s,e}$ | Asset $j$'s gain divided by total beginning portfolio value | decimal return contribution |

Start with shares and prices, rather than assuming a weighted-return formula:

$$
V_{j,s}=N_jP_{j,s},\qquad V_{j,e}=N_jP_{j,e},
\qquad W_s=\sum_{j=1}^{J}V_{j,s}>0,
\qquad W_e=\sum_{j=1}^{J}V_{j,e}.
$$

Beginning weights divide each beginning holding value by the same total:

$$
w_{j,s}=\frac{V_{j,s}}{W_s},\qquad
\sum_{j=1}^{J}w_{j,s}=1.
$$

An asset's gain can now be written as its beginning value times its own return:

$$
\begin{aligned}
V_{j,e}-V_{j,s}
&=N_j(P_{j,e}-P_{j,s})\\
&=N_jP_{j,s}\left(\frac{P_{j,e}}{P_{j,s}}-1\right)
=V_{j,s}r_{j;s,e}.
\end{aligned}
$$

Add those gains and divide by the total **beginning** value:

$$
\begin{aligned}
r_{p;s,e}
&=\frac{W_e-W_s}{W_s}\\
&=\frac{\sum_{j=1}^{J}(V_{j,e}-V_{j,s})}{W_s}\\
&=\sum_{j=1}^{J}\frac{V_{j,s}}{W_s}r_{j;s,e}
=\sum_{j=1}^{J}w_{j,s}r_{j;s,e}.
\end{aligned}
$$

The weights are therefore not an extra adjustment to the answer. They record how much beginning capital produced each asset's gain. The return contribution is

$$
c_{j;s,e}=\frac{V_{j,e}-V_{j,s}}{W_s}
=w_{j,s}r_{j;s,e},\qquad
\sum_{j=1}^{J}c_{j;s,e}=r_{p;s,e}.
$$

A contribution of 0.01 adds 1 percentage point to portfolio return. It is not necessarily a 1% return on that asset, nor 1% of total portfolio profit. For two assets, the last identity is simply $r_p=w_Dr_D+w_Er_E$ over their common interval.

[Zivot, Section 1.3, Portfolios and Portfolio Returns](https://bookdown.org/compfinezbook/introFinRbook/Return-Calculations.html#portfolios-and-portfolio-returns) supports beginning-capital weighting and return contributions. The examples below use our own teaching inputs and calculations. Weights are given, not optimized; method selection begins in Week 8.

### Why not use the ending weights for the interval that just finished?

Ending weights describe where the money is after returns have occurred. They need not equal the beginning capital fractions, even with no trades. Let $h_{j,e}$ be asset $j$'s decimal weight at the end, before any new trade. Under Code 2's conditions,

$$
\begin{aligned}
h_{j,e}&=\frac{V_{j,e}}{W_e}\\
&=\frac{W_sw_{j,s}(1+r_{j;s,e})}{W_s(1+r_{p;s,e})}
=\frac{w_{j,s}(1+r_{j;s,e})}{1+r_{p;s,e}}.
\end{aligned}
$$

An asset whose return exceeds the portfolio return gains weight when its beginning weight is positive. Its share quantity has not increased. Its value has increased more, or fallen less proportionally, than total portfolio value. Multiplying the completed interval's returns by these ending weights generally will not reproduce the gain divided by beginning wealth. That calculation uses the allocation produced by the completed interval's gains and losses, not the allocation that was held at its beginning.

Ending weights can become the **next** interval's beginning weights if no trade or cash flow intervenes. They do not automatically reset to 70/30. This distinction explains the dates attached to weights; it does not yet specify or compare rebalancing strategies.

### Why can we weight simple returns, but not directly weight log returns?

The portfolio has one beginning value and one ending value. Its log return must describe the ratio of those **total values**, not the average of logarithms calculated separately for each asset. Keep the same holdings, prices and no-trade/no-distribution conditions used above.

Let $\ell_j=\ln(1+r_j)$ denote an asset's decimal log return over the common interval, and let $\ell_p$ denote the portfolio's decimal log return. Here $r_j$ and $r_p$ are the corresponding simple returns; $w_j$ is a beginning value weight. All gross returns $1+r_j$ and both portfolio values must be positive. The endpoint subscripts are omitted only to keep the following equations short.

The previous value calculation gives

$$
\frac{W_e}{W_s}=1+r_p
=1+\sum_j w_j r_j
=\sum_j w_j(1+r_j),\qquad \sum_j w_j=1.
$$

Therefore the portfolio log return is

$$
\ell_p=\ln\left(\frac{W_e}{W_s}\right)
=\ln\left(\sum_j w_j(1+r_j)\right)
=\ln\left(\sum_j w_j e^{\ell_j}\right).
$$

Taking the logarithm after adding the weighted gross returns is generally different from adding their weighted logarithms. Exponentiating the latter gives

$$
\exp\left(\sum_j w_j\ell_j\right)
=\prod_j(1+r_j)^{w_j},
$$

a weighted geometric mean of gross returns, not the sum of ending holding values divided by beginning wealth. The two calculations agree if all assets with positive weight have the same return, but not in general. [Zivot, Section 1.3.2](https://bookdown.org/compfinezbook/introFinRbook/Return-Calculations.html#continuously-compounded-portfolio-returns) states this portfolio-return distinction.

This does not contradict Issue 9. Log returns over **consecutive intervals of the same wealth series** add because the intermediate wealth values cancel in a product of ratios. Here we combine **different assets over the same interval**, which requires adding their money values first. A small numerical difference in one example does not turn the incorrect aggregation into an identity.

### How to calculate the required rebalancing

Issue 15 showed why asset weights can change while share quantities stay fixed. Rebalancing means buying or selling to restore the stated weights. Let $V_{j,e}$ be asset $j$'s value immediately before the trade and $W_e=\sum_j V_{j,e}$ the total value. Let $w_j$ be its required weight after the trade, with nonnegative weights summing to one. The desired holding value and trade amount are

$$
\text{desired value}_j=w_jW_e,\qquad
\text{trade amount}_j=w_jW_e-V_{j,e}.
$$

A positive trade amount is a purchase; a negative one is a sale. All three values use the same money units, or the same normalized wealth units. Under the stated zero-cost assumptions,

$$
\sum_j\text{trade amount}_j
=W_e\sum_jw_j-\sum_jV_{j,e}=0.
$$

Thus purchases are funded by sales, without an external deposit. Total value is unchanged by the rebalance itself. The new holdings, not the trades, earn the next interval's return. [Zivot, Section 1.3.1.2](https://bookdown.org/compfinezbook/introFinRbook/Return-Calculations.html#constant-portfolio-weights) explains why restoring weights generally requires trades.

### Predict the short example before execution

This is a separate artificial example, not the F/G case solution. Two holdings are worth 84 and 26 currency units immediately before a trade, and the prescribed target is 75%/25%. Predict the total value, both desired values and both signed trade amounts. Check whether the trades sum to zero and whether total value changes.


````python
# Separate worked example: value transfers needed to restore given weights.
rebalance_before = pd.Series({"Artificial asset X": 84.0, "Artificial asset Y": 26.0})
rebalance_target = pd.Series({"Artificial asset X": 0.75, "Artificial asset Y": 0.25})
rebalance_total = rebalance_before.sum()
rebalance_after = rebalance_target * rebalance_total
rebalance_trades = rebalance_after - rebalance_before
rebalance_example = pd.DataFrame({
    "Before trade (currency units)": rebalance_before,
    "Target weight (%)": 100 * rebalance_target,
    "Desired value (currency units)": rebalance_after,
    "Buy (+) / sell (−), currency units": rebalance_trades,
})
display(rebalance_example.style.format("{:.2f}").set_caption(
    "Separate artificial 75/25 rebalancing example; fixed prices during trading, "
    "no costs, taxes or external cash flows; not the Case 4 solution"
))
assert np.isclose(rebalance_trades.sum(), 0.0)
assert np.isclose(rebalance_after.sum(), rebalance_total)
assert np.allclose(rebalance_after / rebalance_total, rebalance_target)
````
